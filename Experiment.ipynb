```python
import pandas as pd

# Load data
data = pd.read_csv("enjoysport.csv")
print("Training data:")
data
```

```python
def find_s_algorithm(data):
    attributes = data.columns[:-1]  # all columns except target
    target = data.columns[-1]

    # Step 1: Initialize hypothesis with first positive example
    hypothesis = None
    for i in range(len(data)):
        if data.loc[i, target] == "Yes":
            hypothesis = list(data.loc[i, attributes])
            break

    # Step 2: Refine hypothesis using remaining positive examples
    for i in range(len(data)):
        if data.loc[i, target] == "Yes":
            for j in range(len(attributes)):
                if hypothesis[j] != data.loc[i, attributes[j]]:
                    hypothesis[j] = '?'

    return hypothesis

final_hypothesis = find_s_algorithm(data)
print("Most specific hypothesis found by FIND-S:")
print(final_hypothesis)
```
